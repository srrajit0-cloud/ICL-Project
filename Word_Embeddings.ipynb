# Word Embeddings Demo

Compare one-hot vectors, which only identify vocabulary entries, with Word2Vec vectors learned from context. Tiny toy corpora are only for demonstration and do not produce useful general-purpose embeddings.


```python
# Install once if needed: %pip install scikit-learn gensim scipy
from sklearn.preprocessing import OneHotEncoder
from gensim.models import Word2Vec

words = [["cat"], ["dog"], ["apple"], ["car"]]
encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
one_hot = encoder.fit_transform(words)
print("Vocabulary:", encoder.categories_[0])
print("One-hot matrix:\n", one_hot)
print("Shape:", one_hot.shape)
```

```python
sentences = [
    ["python", "is", "a", "language"],
    ["java", "is", "a", "language"],
    ["python", "is", "easy"],
    ["java", "is", "easy"],
    ["python", "is", "powerful"],
    ["java", "is", "powerful"],
 ]
model = Word2Vec(
    sentences=sentences, vector_size=10, window=2, min_count=1,
    workers=1, seed=42, epochs=100
 )
print("Words similar to python:", model.wv.most_similar("python", topn=3))
print("Python vector:", model.wv["python"])
print("Vocabulary size:", len(model.wv.key_to_index))
```

## Takeaways

- One-hot encoding creates distinct binary IDs and does not encode meaning.
- Word2Vec learns context-based vectors from a corpus.
- Sentence-transformer embeddings are often more appropriate for semantic document retrieval.
- Try a representative corpus, compare embedding models, and visualize vectors with PCA for further study.
