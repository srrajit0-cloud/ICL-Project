# Calculating Cosine Similarity Mathematically

Cosine similarity measures the angle between two non-zero vectors: `A · B / (||A|| ||B||)`. It is commonly used to compare text embeddings. Similarity scores are signals, not proof that two passages mean the same thing.


```python
# Install once if needed: %pip install numpy scikit-learn
import math
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

vector_a = [2, 4, 1]
vector_b = [1, 3, 5]

def dot_product(a, b):
    if len(a) != len(b):
        raise ValueError("Vectors must have equal dimensions")
    return sum(x * y for x, y in zip(a, b))

def magnitude(vector):
    return math.sqrt(sum(value * value for value in vector))

def cosine_similarity_manual(a, b):
    denominator = magnitude(a) * magnitude(b)
    return 0.0 if denominator == 0 else dot_product(a, b) / denominator

manual = cosine_similarity_manual(vector_a, vector_b)
numpy_value = float(np.dot(vector_a, vector_b) / (np.linalg.norm(vector_a) * np.linalg.norm(vector_b)))
sklearn_value = float(cosine_similarity(np.array(vector_a).reshape(1, -1), np.array(vector_b).reshape(1, -1))[0, 0])
print("Dot product:", dot_product(vector_a, vector_b))
print("Manual:", round(manual, 6))
print("NumPy:", round(numpy_value, 6))
print("scikit-learn:", round(sklearn_value, 6))
assert np.isclose(manual, numpy_value) and np.isclose(manual, sklearn_value)
```

```python
examples = {
    "Identical vectors": ([1, 2, 3], [1, 2, 3]),
    "Opposite vectors": ([1, 2, 3], [-1, -2, -3]),
    "Perpendicular vectors": ([1, 0], [0, 1]),
    "Scaled vectors": ([1, 2, 3], [2, 4, 6]),
 }
for label, (a, b) in examples.items():
    print(f"{label:24s} -> {cosine_similarity_manual(a, b):.4f}")
```

## Interpretation

For the usual cosine formula, identical directions have similarity 1, perpendicular vectors 0, and opposite directions -1. Zero vectors need special handling because their magnitude makes the denominator zero. In vector databases, metric conventions and embedding normalization can affect distance interpretation.
